```python
categorical_columns = [
    "City",
    "Policy_Type",
    "Policy_Status",
    "Claim_Status",
    "Sales_Channel",
    "Payment_Mode"
]

for col in categorical_columns:
    df[col] = df[col].fillna(
        df[col].mode()[0]
    )
```

```python
Q1 = df["Annual_Premium"].quantile(0.25)
Q3 = df["Annual_Premium"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Annual_Premium"] < lower_limit) |
    (df["Annual_Premium"] > upper_limit)
]

print("Premium outliers:", len(outliers))
```

```python
df["Premium_Outlier_Flag"] = np.where(
    (df["Annual_Premium"] < lower_limit) |
    (df["Annual_Premium"] > upper_limit),
    "Yes",
    "No"
)
```

```python
df["Claim_Ratio"] = np.where(
    df["Annual_Premium"] > 0,
    df["Claim_Amount"] / df["Annual_Premium"],
    0
)
```

```python
age_mapping = {
    "18-25": 1,
    "26-35": 2,
    "36-45": 3,
    "46-55": 4,
    "56-65": 5,
    "65+": 6
}

df["Age_Group_Code"] = df["Age_Group"].map(age_mapping)
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "65+"
    ]
)
```

```python
df["Premium_Band"] = pd.cut(
    df["Annual_Premium"],
    bins=[0, 10000, 25000, 50000, 100000, np.inf],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High",
        "Premium"
    ]
)
```
